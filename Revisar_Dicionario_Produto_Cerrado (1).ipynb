## Funções de aplicação
A leitura preserva identificadores, zeros iniciais, vírgulas decimais e campos vazios. Nas vistas de apoio, as seis colunas auxiliares inadequadas recebem prefixo LEGADO_; campos novos usam fontes explícitas. Em assentamentos e quilombolas não é inventada uma modalidade territorial ausente.

As vistas e os recortes UC/ZA são derivados dos mesmos registros: **não os concatene aos CSVs originais**. A soma de áreas apresentada é acumulada por linhas, não área física de união.


```python
from pathlib import Path
from datetime import datetime, timezone
import hashlib, json, shutil, zipfile, html
import pandas as pd

def sha256(p):
    h=hashlib.sha256()
    with Path(p).open('rb') as f:
        for b in iter(lambda:f.read(1024*1024),b''):h.update(b)
    return h.hexdigest()

def ler(p):
    return pd.read_csv(p, sep=';', dtype=str, keep_default_na=False, encoding='utf-8-sig')

def salvar(df,p):
    df.to_csv(p,sep=';',index=False,encoding='utf-8-sig')

def limpar(s):
    return s.astype(str).str.strip().str.replace(r'\.0$','',regex=True).replace({'':'Não informado','nan':'Não informado','None':'Não informado','<NA>':'Não informado','NULL':'Não informado'})

def numero(s):
    # Exportação original: pontos OU vírgulas decimais, sem separador de milhar.
    return pd.to_numeric(s.str.strip().str.replace(',','.',regex=False),errors='coerce')

# Mapeamento explícito: nenhuma busca por substring.
ACOES = [
 ('quilombolas','_TIPOLOGIA_2','TIPO_IMPEDIMENTO','ROTULO_CAMADA_REV','TIPO_IMPEDIMENTO','Rótulo da camada; modalidade territorial não disponível nesta coluna.'),
 ('assentamentos','_TIPOLOGIA_2','TIPO_PRODUTOR_ANALISE','TIPO_PRODUTOR_REV','TIPO_PRODUTOR_ANALISE','Classificação do produtor. Não representa modalidade do assentamento.'),
 ('embargos','_TIPOLOGIA_3','TIPO_PRODUTOR_ANALISE','TIPO_DANO_EMBARGO_REV','EMB_TIPO_AREA','Usar o tipo de dano do embargo; a coluna antiga contém porte/classificação do produtor.'),
 ('embargos','_UF_UNIDADE','UF_OPERACAO','UF_EMBARGO_REV','EMB_UF','Usar a UF do local embargado para caracterização da camada; conservar UF_OPERACAO para análise do crédito.'),
 ('car_suspenso','_TIPOLOGIA_2','TIPO_PRODUTOR_ANALISE','TIPO_IMOVEL_CAR_REV','CAR_TIPO_IMOVE','Usar o tipo cadastral do imóvel. IRU não significa pequeno, médio ou grande produtor.'),
 ('car_suspenso','_UF_UNIDADE','UF_OPERACAO','UF_IMOVEL_CAR_REV','CAR_UF','Usar a UF do imóvel na camada CAR; UF_OPERACAO continua representando a operação.')
]

def aplicar_revisao_v2(pasta,pacote,destino_pai=None):
    pasta=Path(pasta)
    if not pasta.is_dir():
        raise FileNotFoundError(f'Pasta não encontrada: {pasta}. Corrija PASTA_ORIGINAL; use a pasta de documentação de 02:46, não a primeira pasta de 02:15.')
    for nome in ['dicionario_colunas.csv','definicoes_pendentes.csv','status.json','validacao.csv','inventario_origens.json','alertas_campos_auxiliares.csv','LEIA_ME.md']:
        if not (pasta/nome).is_file():raise FileNotFoundError(f'Arquivo ausente: {nome}')
    if sha256(pasta/'definicoes_pendentes.csv')!=pacote['sha256_pendencias']:
        raise ValueError('As pendências diferem das 149 entradas revisadas. Não aplicar este pacote a outra versão.')
    dados={}
    for nome,h in pacote['sha256_dados_esperados'].items():
        if Path(nome).name!=nome:raise ValueError('Nome de arquivo inválido no pacote.')
        arq=pasta/'dados'/nome
        if not arq.is_file() or sha256(arq)!=h:raise ValueError(f'{nome}: conteúdo diferente da versão revisada; aplicação interrompida.')
        dados[Path(nome).stem]=ler(arq)
    if len(dados)!=9:raise ValueError('São necessários os nove CSVs originais.')
    valid=ler(pasta/'validacao.csv')
    if 'confere' not in valid or not valid.confere.str.lower().eq('true').all():raise ValueError('As verificações de origem não passaram.')
    atual=ler(pasta/'dicionario_colunas.csv'); pend=ler(pasta/'definicoes_pendentes.csv')
    regras=pd.DataFrame(pacote['regras'])
    chaves=['conjunto','coluna']
    if atual.duplicated(chaves).any() or regras.duplicated(chaves).any():raise ValueError('Chaves duplicadas no dicionário/revisão.')
    if set(map(tuple,pend[chaves].values))!=set(map(tuple,regras[chaves].values)):raise ValueError('Regras não cobrem exatamente as pendências.')
    indice={tuple(row):i for i,row in enumerate(atual[chaves].values)}
    diagnostico=[]
    for r in regras.to_dict('records'):
        key=(r['conjunto'],r['coluna']);df=dados[r['conjunto']]
        if key not in indice or r['coluna'] not in df:raise ValueError(f'Campo ausente: {key}')
        s=df[r['coluna']]; nonempty=s.str.strip().ne('')
        diagnostico.append({'conjunto':r['conjunto'],'coluna':r['coluna'],'registros':len(s),'vazios':int((~nonempty).sum()),'valores_distintos_nao_vazios':int(s[nonempty].nunique()),'situacao_revisao_2':r['situacao_revisao_2']})
        i=indice[key]
        for c in ['descricao','unidade','tipo_logico','fonte_definicao','localizacao_fonte','situacao_revisao_2','observacao_revisao_2']:
            atual.loc[i,c]=r[c]
        atual.loc[i,'situacao_definicao']='definição documentada na revisão 2' if r['situacao_revisao_2']=='documentada' else 'definição pendente'
        atual.loc[i,'criterio_revisao']='Metadados oficiais confrontados com nomes e conteúdo da versão fixada por SHA-256; consulta '+pacote['data_consulta']
        atual.loc[i,'observacao_revisao']=r['observacao_revisao_2']
    atual=atual.fillna('')
    tratamento=[]; vistas={}; dic_vistas=[]; resumo_dim=[]
    for conjunto,legado,derivadora,novo,fonte,obs in ACOES:
        df=dados[conjunto]
        if any(c not in df for c in [legado,derivadora,fonte]):raise ValueError(f'Campo necessário ausente em {conjunto}.')
        n=int(df[legado].ne(limpar(df[derivadora])).sum())
        if n:raise ValueError(f'O alerta {conjunto}/{legado} não corresponde à derivação prevista ({n} divergências).')
        destino_legado='LEGADO_'+legado.lstrip('_')
        if conjunto not in vistas:vistas[conjunto]=df.copy()
        if novo in df or destino_legado in df:raise ValueError('A entrada já contém nomes reservados da revisão.')
        vistas[conjunto]=vistas[conjunto].rename(columns={legado:destino_legado})
        vistas[conjunto][novo]=df[fonte] # cópia textual exata, inclusive vazios
        dic_vistas += [
            {'arquivo':conjunto+'_campos_revisados.csv','coluna':destino_legado,'origem':legado,'descricao':'Campo legado derivado de '+derivadora+'. '+obs},
            {'arquivo':conjunto+'_campos_revisados.csv','coluna':novo,'origem':fonte,'descricao':obs+' Cópia textual exata de '+fonte+'.'}]
        tratamento.append({'conjunto':conjunto,'campo_legado':legado,'derivadora_confirmada':derivadora,'divergencias_derivacao':n,'campo_recomendado':novo,'fonte_exata':fonte,'linhas_diferentes_entre_legado_e_recomendado':int(limpar(df[legado]).ne(limpar(df[fonte])).sum()),'fonte_vazia':int(df[fonte].str.strip().eq('').sum()),'tratamento':obs,'situacao':'tratado na documentação e na vista de apoio; relatórios anteriores não alterados'})
        i=indice[(conjunto,legado)]
        atual.loc[i,'descricao']='Campo legado derivado de '+derivadora+'. '+obs
        atual.loc[i,'observacao_revisao']=obs+' Na vista de apoio, passa a '+destino_legado+'; campo recomendado: '+novo+'.'
        for valor,t in df.groupby(fonte,dropna=False):
            resumo_dim.append({'conjunto':conjunto,'campo':novo,'valor':valor or '(vazio)','registros':len(t),'glebas_distintas':t['ID_GLEBA2'].nunique(),'operacoes_distintas':t['REF_BACEN'].nunique()})
    # Conferências semânticas adicionais de campos documentados.
    car=dados['car_suspenso']
    if not car.CAR_STATUS_CAR_CODIGO.eq(car.CAR_STATUS_IMO).all():raise ValueError('Código CAR diverge de CAR_STATUS_IMO.')
    if not set(car.CAR_STATUS_CAR_CODIGO.unique())<= {'SU'}:raise ValueError('CAR apresenta status não previsto nesta versão.')
    # UC e zona de amortecimento: contagens e soma administrativa, nunca área de união.
    uc=dados['unidades_conservacao']; lim=uc.UC_LIMITE.str.strip().str.lower()
    if not lim.isin(['uc','za']).all():raise ValueError('UC_LIMITE contém códigos não documentados.')
    rel_uc=[]
    for codigo,t in uc.groupby(lim):
        a=numero(t.AREA_INTERSECAO_HA)
        if a.isna().any():raise ValueError('Área da interseção de UC não numérica; resumo interrompido.')
        rel_uc.append({'limite':codigo,'significado':'limites da unidade de conservação' if codigo=='uc' else 'zona de amortecimento','registros':len(t),'glebas_distintas':t.ID_GLEBA2.nunique(),'operacoes_distintas':t.REF_BACEN.nunique(),'unidades_cnuc_distintas':t.UC_CD_CNUC.nunique(),'area_acumulada_ha':float(a.sum()),'nota':'Soma das linhas; inclui repetições. Não é área da união. Contagens distintas entre grupos podem se sobrepor.'})
    perfil=pd.DataFrame(diagnostico); rem=atual[atual.situacao_definicao.eq('definição pendente')].copy()
    stamp=datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S_%fZ')
    pai=Path(destino_pai) if destino_pai else pasta.parent
    pai.mkdir(parents=True,exist_ok=True)
    nova=pai/('produto_Cerrado_revisao_2_'+stamp)
    shutil.copytree(pasta,nova)
    hist=nova/'historico_revisao_2';hist.mkdir()
    for nome in ['dicionario_colunas.csv','definicoes_pendentes.csv','status.json','LEIA_ME.md','alertas_campos_auxiliares.csv']:
        shutil.copy2(nova/nome,hist/nome)
    salvar(atual,nova/'dicionario_colunas.csv');salvar(rem,nova/'definicoes_pendentes.csv')
    salvar(regras,nova/'revisao_149_definicoes.csv');salvar(perfil,nova/'perfil_149_campos.csv')
    salvar(pd.DataFrame(tratamento),nova/'tratamento_seis_alertas.csv')
    alertas_anteriores=ler(nova/'alertas_campos_auxiliares.csv')
    alertas_anteriores['situacao_atual']='Tratado na revisão 2; consultar tratamento_seis_alertas.csv. Relatórios anteriores ainda requerem revisão.'
    salvar(alertas_anteriores,nova/'alertas_campos_auxiliares.csv')
    apoio=nova/'apoio_revisao';apoio.mkdir()
    for g,df in vistas.items():
        if len(df)!=len(dados[g]):raise AssertionError('Contagem alterada na vista de apoio.')
        # Cada campo não afetado continua idêntico, incluindo a ordem dos registros.
        afetados={a[1] for a in ACOES if a[0]==g}
        comuns=[c for c in dados[g] if c not in afetados]
        if not df[comuns].equals(dados[g][comuns]):raise AssertionError('Campo original alterado na vista.')
        salvar(df,apoio/(g+'_campos_revisados.csv'))
    salvar(pd.DataFrame(dic_vistas),apoio/'dicionario_campos_revisados.csv')
    salvar(pd.DataFrame(resumo_dim),apoio/'contagens_dimensoes_revisadas.csv')
    salvar(uc.loc[lim.eq('uc')],apoio/'unidades_conservacao_limites_uc.csv')
    salvar(uc.loc[lim.eq('za')],apoio/'unidades_conservacao_zonas_amortecimento.csv')
    salvar(pd.DataFrame(rel_uc),apoio/'resumo_uc_e_za.csv')
    fontes=[{'id':k,'titulo':v[0],'url':v[1],'data_consulta':pacote['data_consulta']} for k,v in pacote['fontes'].items()]
    salvar(pd.DataFrame(fontes),nova/'fontes_revisao_2.csv')
    (nova/'regras_revisao_2.json').write_text(json.dumps(pacote,ensure_ascii=False,indent=2),encoding='utf-8')
    for nome,h in pacote['sha256_dados_esperados'].items():
        if sha256(nova/'dados'/nome)!=h:raise AssertionError('Dados originais alterados; ZIP não produzido.')
    old=json.loads((pasta/'status.json').read_text(encoding='utf-8'))
    status=dict(old)
    status.update({'versao_revisao_documental':'2.0','revisao_documental_utc':stamp,'definicoes_pendentes':len(rem),'entradas_documentadas_nesta_revisao':int(regras.situacao_revisao_2.eq('documentada').sum()),'entradas_parciais_nesta_revisao':int(regras.situacao_revisao_2.eq('parcial').sum()),'entradas_sem_definicao_suficiente':int(regras.situacao_revisao_2.eq('pendente').sum()),'alertas_campos_auxiliares':6,'alertas_auxiliares_tratados_no_produto':6,'alertas_novos_relevantes':1,'registros_uc':int(lim.eq('uc').sum()),'registros_zona_amortecimento':int(lim.eq('za').sum()),'status':'REVISAO_2_COM_PENDENCIAS_DOCUMENTAIS_E_RECORTE_UC_ZA','dados_modificados':False,'vistas_de_apoio_criadas':True,'intersecoes_recalculadas':False,'relatorios_da_dissertacao_atualizados':False,'observacao_status':'As 149 entradas foram examinadas. Definições parciais permanecem pendentes. Seis alertas tratados em dicionário e vistas de apoio. Separação UC/ZA exige rever os resumos correspondentes da dissertação. Não houve verificação administrativa.'})
    (nova/'status.json').write_text(json.dumps(status,ensure_ascii=False,indent=2),encoding='utf-8')
    log={'origem':str(pasta),'destino':str(nova),'sha256_dados_originais':pacote['sha256_dados_esperados'],'pendencias_antes':len(pend),'pendencias_depois':len(rem),'status_anterior':old,'versao':'2.0'}
    (nova/'registro_revisao_2.json').write_text(json.dumps(log,ensure_ascii=False,indent=2),encoding='utf-8')
    intro=f'''# Revisão 2 do produto Cerrado\n\nRevisão das 149 definições restantes: {status['entradas_documentadas_nesta_revisao']} documentadas, {status['entradas_parciais_nesta_revisao']} parciais e {status['entradas_sem_definicao_suficiente']} sem definição suficiente. Total ainda pendente: {len(rem)}.\n\nOs nove CSVs em dados/ foram preservados byte a byte. dicionario_colunas.csv é a documentação vigente. As fontes, decisões e observações constam em revisao_149_definicoes.csv. Os arquivos de apoio são derivados dos mesmos registros e NÃO devem ser concatenados aos originais, pois isso duplicaria ocorrências.\n\n## Seis alertas\nOs nomes parciais usados nos notebooks anteriores selecionavam colunas inadequadas. Os arquivos *_campos_revisados.csv usam nomes explícitos; as colunas problemáticas recebem prefixo LEGADO_. Em assentamentos não se inventa modalidade a partir do porte do produtor. Em quilombolas a tipologia 2 representa apenas a camada. EMB_TIPO_AREA representa tipo de dano, não a descrição completa da infração. A UF do crédito e a UF da área têm funções distintas. Consulte tratamento_seis_alertas.csv e o dicionário em apoio_revisao/.\n\n## Unidades de conservação e zonas de amortecimento\nO CSV contém {int(lim.eq('uc').sum())} registros UC e {int(lim.eq('za').sum())} registros ZA. O dicionário MMA distingue esses códigos. As duas seleções estão em apoio_revisao/. resumo_uc_e_za.csv informa contagens e área acumulada por linhas. Não calcula área física de união, pois os CSVs leves não contêm a geometria necessária. Contagens distintas de operações/glebas não devem ser somadas entre grupos. Para atualizar mapas ou área da união, filtrar UC_LIMITE na base geométrica e executar somente os resumos afetados.\n\n## O que revisar na dissertação\nConferir tabelas e textos por tipo de imóvel CAR, tipo de infração/embargo, modalidade de assentamento, tipologia quilombola e UF da área. Conferir a seção de UC e a síntese quando o total incluir ZA. A correção do produto não atualiza automaticamente o Word nem os gráficos anteriores.\n\n## Pendências\nAs 36 definições parciais e uma não confirmada continuam em definicoes_pendentes.csv, com a dúvida exata registrada. Incluem unidades físicas não explicitadas, códigos/truncamentos e inconsistências entre metadados e arquivos. Um campo vazio não foi preenchido por suposição. Metadados consultados agora descrevem o significado; não substituem a versão histórica da base nem comprovam sua situação na emissão do crédito.\n\n## Execução\nO notebook é autocontido e usa pandas (disponível no Colab). Aplica regras revisadas, não interpreta automaticamente qualquer base nova. Verifica os hashes da versão analisada e cria uma nova pasta. Para usar outra versão, revisar as regras e os arquivos antes de alterar os controles.\n'''
    (nova/'LEIA_ME.md').write_text(intro,encoding='utf-8')
    # Relatório HTML autônomo: pode ser aberto no navegador ou impresso.
    h='<html lang="pt-br"><meta charset="utf-8"><title>Revisão do produto Cerrado</title><style>body{font:16px Arial;max-width:1250px;margin:40px auto;color:#21332d}h1,h2{color:#245843}table{border-collapse:collapse;width:100%;font-size:13px;margin:20px 0}td,th{border:1px solid #cdd7d1;padding:8px;text-align:left;vertical-align:top;overflow-wrap:anywhere}th{background:#e9f1eb}p{line-height:1.6}.aviso{background:#fff3da;padding:18px}a{color:#245843}@media print{body{margin:0}thead{display:table-header-group}tr{break-inside:avoid}}</style><body>'
    h+='<h1>Revisão do produto Cerrado</h1><p>149 entradas examinadas: '+str(status['entradas_documentadas_nesta_revisao'])+' documentadas; 36 parciais; 1 sem definição suficiente. Nove CSVs originais preservados. Seis alertas tratados nas vistas de apoio.</p>'
    h+='<p class="aviso">Foram identificados 124 registros de zona de amortecimento na camada de unidades de conservação. A separação UC/ZA afeta a interpretação dos resumos territoriais. Este relatório não atualiza o Word da dissertação nem recalcula a união geométrica.</p>'
    h+='<h2>UC e ZA</h2>'+pd.DataFrame(rel_uc).to_html(index=False,escape=True)
    h+='<h2>Tratamento dos seis alertas</h2>'+pd.DataFrame(tratamento).to_html(index=False,escape=True)
    h+='<h2>Definições que ainda precisam de confirmação</h2>'+regras.loc[regras.situacao_revisao_2.ne('documentada'),['conjunto','coluna','descricao','unidade','observacao_revisao_2']].to_html(index=False,escape=True)
    h+='<h2>Revisão completa das 149 entradas</h2>'+regras[['conjunto','coluna','descricao','unidade','situacao_revisao_2','fontes_ids']].to_html(index=False,escape=True)
    h+='<h2>Fontes</h2><ul>'+''.join('<li>'+html.escape(k)+': <a href="'+html.escape(v[1],quote=True)+'">'+html.escape(v[0])+'</a></li>' for k,v in pacote['fontes'].items())+'</ul></body></html>'
    (nova/'Relatorio_Revisao.html').write_text(h,encoding='utf-8')
    zpath=Path(str(nova)+'.zip')
    with zipfile.ZipFile(zpath,'w',zipfile.ZIP_DEFLATED) as z:
        for p in sorted(nova.rglob('*')):
            if p.is_file():z.write(p,p.relative_to(nova))
    print(json.dumps(status,ensure_ascii=False,indent=2))
    return nova,zpath
```
