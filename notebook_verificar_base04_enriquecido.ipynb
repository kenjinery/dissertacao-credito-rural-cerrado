# Verificação da base enriquecida do Base04

Este notebook lê o arquivo `sicor_cerrado_2021_2025_enriquecido_detalhado.csv`, conta seus registros e lista todas as colunas, sem modificar o arquivo original.

```python
# 1. MONTAR O GOOGLE DRIVE
from google.colab import drive
drive.mount('/content/drive')
```

```python
# 2. IMPORTAÇÕES E CAMINHO DO ARQUIVO
from pathlib import Path
import csv
import pandas as pd
from IPython.display import display

ARQUIVO = Path(
    '/content/drive/MyDrive/Colab Notebooks/fluxo/Envio/'
    'Arquivos gerados e relatorios/Base04_enriquecimento_detalhado/'
    'sicor_cerrado_2021_2025_enriquecido_detalhado.csv'
)

if not ARQUIVO.exists():
    raise FileNotFoundError(f'Arquivo não encontrado: {ARQUIVO}')

print('Arquivo localizado:')
print(ARQUIVO)
print(f'Tamanho: {ARQUIVO.stat().st_size / (1024 ** 2):,.2f} MB')
```

```python
# 3. IDENTIFICAR O SEPARADOR E LISTAR AS COLUNAS
with ARQUIVO.open('r', encoding='utf-8-sig', newline='') as arquivo_texto:
    primeira_linha = arquivo_texto.readline()

try:
    SEPARADOR = csv.Sniffer().sniff(primeira_linha, delimiters=';,\t|').delimiter
except csv.Error:
    SEPARADOR = ';'

cabecalho = pd.read_csv(
    ARQUIVO,
    sep=SEPARADOR,
    encoding='utf-8-sig',
    nrows=0
)

colunas = cabecalho.columns.tolist()
tabela_colunas = pd.DataFrame({
    'Nº': range(1, len(colunas) + 1),
    'COLUNA': colunas
})

print(f'Separador identificado: {repr(SEPARADOR)}')
print(f'Quantidade de colunas: {len(colunas):,}')
print('\nLista de colunas:')
display(tabela_colunas)
```

```python
# 4. CONTAR OS REGISTROS EM BLOCOS
# A leitura em blocos evita carregar toda a base na memória de uma só vez.
TAMANHO_BLOCO = 100_000
total_registros = 0

leitor = pd.read_csv(
    ARQUIVO,
    sep=SEPARADOR,
    encoding='utf-8-sig',
    usecols=[colunas[0]],
    chunksize=TAMANHO_BLOCO,
    low_memory=False
)

for numero_bloco, bloco in enumerate(leitor, start=1):
    total_registros += len(bloco)
    print(
        f'Bloco {numero_bloco:03d}: {len(bloco):,} registros | '
        f'acumulado: {total_registros:,}'
    )

print('\n' + '=' * 60)
print('RESUMO DA BASE')
print('=' * 60)
print(f'Quantidade total de registros: {total_registros:,}')
print(f'Quantidade total de colunas:   {len(colunas):,}')
```

```python
# 5. EXIBIR UMA AMOSTRA DOS REGISTROS (OPCIONAL)
amostra = pd.read_csv(
    ARQUIVO,
    sep=SEPARADOR,
    encoding='utf-8-sig',
    nrows=5,
    low_memory=False
)

display(amostra)
```
