```python
import json
PACOTE_REVISAO = json.loads(r'''{"origem_sha256": "bfe129a3e420719cf12d6f281b381480a5beb59e9c0222ae76a3e0308a31b5c4", "registros_originais": [{"conjunto": "terras_indigenas", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "420", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "420", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_GID", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_TERRAI_COD", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_TERRAI_NOM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_ETNIA_NOME", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_UF_SIGLA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_SUPERFICIE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_FASE_TI", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_MODALIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_REESTUDO_T", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "520", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_CR", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_FAIXA_FRON", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_UNDADM_COD", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_UNDADM_NOM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_UNDADM_SIG", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_DOMINIO_UN", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_DATA_ATUAL", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "2", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_EPSG", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "TI_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_TI_ID", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_AREA_TOTAL_TI_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_TI_NOME", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_TI_FASE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_TI_MODALIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_TI_ETNIA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_TI_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "terras_indigenas", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_CD_QUILOMB", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "225", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_CD_SR", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_PROCESS", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "34", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_NM_COMUNID", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_NM_MUNICIP", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_CD_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_DT_PUBLICA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "35", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_DT_PUBLIC1", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "287", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_FAMILIA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "34", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_DT_TITULAC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "262", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_AREA_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "24", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_PERIMET", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "197", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_CD_SIPRA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "358", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_OB_DESCRIC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "221", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_ST_TITULAD", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "224", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_DT_DECRETO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "237", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_TP_LEVANTA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "305", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_ESCALAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "301", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_AREA_CALC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "35", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_PERIMETRO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "35", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_ESFERA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "34", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_FASE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "34", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_RESPONSAVE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "35", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "QUIL_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "quilombolas", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "3", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "8266", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "3", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "8266", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_NOME", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_CATEGORIA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_PROTECAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_TIPO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_GOVERNO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_COMUNITARI", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_CLASSE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_ORGAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1016", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_SOBREPOSIC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_ATOLEGAL", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "22994", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_ANOCRIACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "3412", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_ESTAGIO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "17901", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_CODIGO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "33984", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_OBSERVACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "33984", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_SHAPE_LENG", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_SHAPE_AREA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_BIOMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_GEOCODIGO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_AREA_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "FP_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "florestas_publicas", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "3756", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "3756", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_CD_SIPRA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_NOME_PROJE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_AREA_HECTA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_CAPACIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_NUM_FAMILI", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_FASE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_DATA_DE_CR", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_FORMA_OBTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "20", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_DATA_OBTEN", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "20", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_AREA_CALC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1188", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_SR", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "33044", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_DESCRICAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "33044", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "ASSENT_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "assentamentos", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "723", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "723", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_OBJECTID", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_SEQ_TAD", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_NUM_TAD", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_SERIE_TAD", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "567", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "884", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_ORIGEM_GEO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_COD_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_COD_MUNICI", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_NOME_IMOVE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "753", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_DES_LOCALI", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "180", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_NOME_EMBAR", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "25", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_CPF_CNPJ_E", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "26", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_SIT_DESMAT", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_TIPO_AREA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "279", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_NUM_AUTO_I", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "133", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_SERIE_AUTO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "594", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_COD_TIPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "303", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_DES_TIPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "303", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_UNID_CONTR", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_ORDEM_FISC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "278", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_CD_ACAO_FI", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "405", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_NUM_PROCES", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "46", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_DES_TAD", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_DES_INFRAC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "215", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_NUM_LONGIT", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_NUM_LATITU", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "13", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_DAT_EMBARG", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_DAT_IMPRES", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "53", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_DAT_ULT_AL", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_NUM_LONG00", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_NUM_LATI00", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "13", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_QTD_AREA_D", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "648", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_QTD_AREA_E", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "56", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_DAT_ULT_00", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_ST_AREA_SH", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_ST_PERIMET", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "EMB_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "embargos", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "5", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "10074", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "5", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "10074", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_GML_ID", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18617", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_DOCLEG_ID", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_UC_ID", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_CD_CNUC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_WDPA_PID", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1667", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_NOME_UC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_CRIA_ANO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_CRIA_ATO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_OUTRO_ATO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_PL_MANEJO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_CO_GESTOR", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_QUALI_POL", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_PPGR", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "14893", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_HA_TOTAL", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "40", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_HA_ATO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_ESFERA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_ORG_GESTOR", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_GRUPO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_CATEGORIA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_CAT_IUCN", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_AMAZONIA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18011", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_CAATINGA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "15324", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_CERRADO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "40", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_MATLANTICA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "15535", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_PAMPA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18617", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_PANTANAL", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18616", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_MARINHO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "17664", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_SITUACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_LIMITE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "UC_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "unidades_conservacao", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "193", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "193", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_COD_IMOVEL", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_STATUS_IMO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_DAT_CRIACA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_DATA_ATUAL", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "34", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_AREA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_CONDICAO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_COD_MUNICI", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_M_FISCAL", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_TIPO_IMOVE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_STATUS_CAR_CODIGO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_UF_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_ARQUIVO_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "CAR_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}, {"conjunto": "car_suspenso", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo preservado literalmente. Consultar o dicionário da fonte antes de interpretar.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados"}], "revisoes": [{"conjunto": "terras_indigenas", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Índice da geometria dentro da ordem, preservado da base territorial.", "tipo_logico": "identificador", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Código da categoria do emitente utilizado no vínculo com o arquivo de categorias.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Campo de estado preservado da base operacional. O código Base04 tenta utilizá-lo no vínculo com siglas de UF; não é uma UF calculada pela interseção.", "tipo_logico": "código de origem", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do programa de crédito utilizado no vínculo com o arquivo Programa.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Área financiada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do subprograma de crédito utilizado no vínculo com o arquivo Subprogramas.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "420", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Área informada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do programa de crédito obtida por junção com a tabela de programas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Descrição da categoria do emitente obtida por junção com a tabela de categorias.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do subprograma obtida por junção com a tabela de subprogramas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "420", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Quantidade de linhas da base operacional agrupadas pela chave REF_BACEN e ordem. Não é número de glebas.", "tipo_logico": "inteiro", "unidade": "linhas", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Classificação analítica por ordem: pequeno/Pronaf, médio/Pronamp, grande ou demais/não classificado. Base04 usa, nessa precedência, categoria do emitente e descrições de categoria e programa; não constitui nova classificação oficial do beneficiário.", "tipo_logico": "categoria analítica", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Sim quando a descrição do subprograma contém GRUPO B (expressão GRUPO\\s*B); Não nos demais casos. É identificação textual realizada por Base04.", "tipo_logico": "indicador textual", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Vinculado quando REF_BACEN e ordem são encontrados na junção com a base operacional; Não vinculado quando não encontrados.", "tipo_logico": "categoria de vínculo", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_GID", "incluida_no_csv_leve": "True", "descricao": "Atributo GID preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_TERRAI_COD", "incluida_no_csv_leve": "True", "descricao": "Atributo TERRAI_COD preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_TERRAI_NOM", "incluida_no_csv_leve": "True", "descricao": "Nome da terra indígena utilizado pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_ETNIA_NOME", "incluida_no_csv_leve": "True", "descricao": "Nome do povo/etnia utilizado pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo MUNICIPIO preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_UF_SIGLA", "incluida_no_csv_leve": "True", "descricao": "Sigla de UF informada na camada de terras indígenas; não calculada pelo notebook de organização.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_SUPERFICIE", "incluida_no_csv_leve": "True", "descricao": "Atributo SUPERFICIE preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_FASE_TI", "incluida_no_csv_leve": "True", "descricao": "Fase da terra indígena utilizada pelo relatório; não reconstituída na data do crédito.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_MODALIDADE", "incluida_no_csv_leve": "True", "descricao": "Modalidade da terra indígena utilizada pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_REESTUDO_T", "incluida_no_csv_leve": "True", "descricao": "Atributo REESTUDO_T preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "520", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_CR", "incluida_no_csv_leve": "True", "descricao": "Atributo CR preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_FAIXA_FRON", "incluida_no_csv_leve": "True", "descricao": "Atributo FAIXA_FRON preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_UNDADM_COD", "incluida_no_csv_leve": "True", "descricao": "Atributo UNDADM_COD preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_UNDADM_NOM", "incluida_no_csv_leve": "True", "descricao": "Atributo UNDADM_NOM preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_UNDADM_SIG", "incluida_no_csv_leve": "True", "descricao": "Atributo UNDADM_SIG preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_DOMINIO_UN", "incluida_no_csv_leve": "True", "descricao": "Atributo DOMINIO_UN preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_DATA_ATUAL", "incluida_no_csv_leve": "True", "descricao": "Atributo DATA_ATUAL preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "2", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_EPSG", "incluida_no_csv_leve": "True", "descricao": "Atributo EPSG preservado da camada de origem com prefixo TI. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Identificador sequencial de feição atribuído no notebook de interseção, iniciado em 1 após a preparação da camada. Depende da versão e ordem da base; não é identificador oficial do órgão.", "tipo_logico": "identificador interno", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "atribuição np.arange(1, len(camada) + 1)", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "TI_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Área da feição completa da camada, calculada no notebook de interseção após projeção EPSG:5880: geometry.area / 10000. Não é área de interseção nem área documental declarada.", "tipo_logico": "numérico de origem", "unidade": "ha", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji/view", "criterio_revisao": "CRS_AREA e cálculo da área da camada", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "terras_indigenas", "coluna": "_TI_ID", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TI_ID_INTERNO no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "TI_ID_INTERNO", "teste_derivacao": "texto"}, {"conjunto": "terras_indigenas", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ANO_BASE no processamento anterior; não é uma nova observação independente. Em terras_indigenas, a origem indicada é ANO_BASE. Não equiparar automaticamente ano do arquivo e ano de emissão.", "tipo_logico": "ano", "unidade": "ano", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ANO_BASE", "teste_derivacao": "numero"}, {"conjunto": "terras_indigenas", "coluna": "_AREA_TOTAL_TI_HA", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TI_AREA_TOTAL_HA no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "ha", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "TI_AREA_TOTAL_HA", "teste_derivacao": "numero"}, {"conjunto": "terras_indigenas", "coluna": "_TI_NOME", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TI_TERRAI_NOM no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "TI_TERRAI_NOM", "teste_derivacao": "texto"}, {"conjunto": "terras_indigenas", "coluna": "_TI_FASE", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TI_FASE_TI no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "TI_FASE_TI", "teste_derivacao": "texto"}, {"conjunto": "terras_indigenas", "coluna": "_TI_MODALIDADE", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TI_MODALIDADE no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "TI_MODALIDADE", "teste_derivacao": "texto"}, {"conjunto": "terras_indigenas", "coluna": "_TI_ETNIA", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TI_ETNIA_NOME no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "TI_ETNIA_NOME", "teste_derivacao": "texto"}, {"conjunto": "terras_indigenas", "coluna": "_TI_UF", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TI_UF_SIGLA no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "TI_UF_SIGLA", "teste_derivacao": "texto"}, {"conjunto": "terras_indigenas", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de _TI_FASE no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "_TI_FASE", "teste_derivacao": "texto"}, {"conjunto": "terras_indigenas", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de _TI_MODALIDADE no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "_TI_MODALIDADE", "teste_derivacao": "texto"}, {"conjunto": "terras_indigenas", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de _TI_ETNIA no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "_TI_ETNIA", "teste_derivacao": "texto"}, {"conjunto": "terras_indigenas", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de _TI_UF no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "660", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "_TI_UF", "teste_derivacao": "texto"}, {"conjunto": "quilombolas", "coluna": "QUIL_CD_QUILOMB", "incluida_no_csv_leve": "True", "descricao": "Atributo CD_QUILOMB preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "225", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_CD_SR", "incluida_no_csv_leve": "True", "descricao": "Atributo CD_SR preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_PROCESS", "incluida_no_csv_leve": "True", "descricao": "Atributo NR_PROCESS preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "34", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_NM_COMUNID", "incluida_no_csv_leve": "True", "descricao": "Atributo NM_COMUNID preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_NM_MUNICIP", "incluida_no_csv_leve": "True", "descricao": "Atributo NM_MUNICIP preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_CD_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo de UF do território quilombola utilizado pelo relatório; conservar a codificação de origem.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1swdqtJoEeWwDhmpmYkPAzP231q6FjBj3/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_DT_PUBLICA", "incluida_no_csv_leve": "True", "descricao": "Atributo DT_PUBLICA preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "35", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_DT_PUBLIC1", "incluida_no_csv_leve": "True", "descricao": "Atributo DT_PUBLIC1 preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "287", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_FAMILIA", "incluida_no_csv_leve": "True", "descricao": "Atributo NR_FAMILIA preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "34", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_DT_TITULAC", "incluida_no_csv_leve": "True", "descricao": "Atributo DT_TITULAC preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "262", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_AREA_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo NR_AREA_HA preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "24", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_PERIMET", "incluida_no_csv_leve": "True", "descricao": "Atributo NR_PERIMET preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "197", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_CD_SIPRA", "incluida_no_csv_leve": "True", "descricao": "Atributo CD_SIPRA preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "358", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_OB_DESCRIC", "incluida_no_csv_leve": "True", "descricao": "Atributo OB_DESCRIC preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "221", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_ST_TITULAD", "incluida_no_csv_leve": "True", "descricao": "Atributo ST_TITULAD preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "224", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_DT_DECRETO", "incluida_no_csv_leve": "True", "descricao": "Atributo DT_DECRETO preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "237", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_TP_LEVANTA", "incluida_no_csv_leve": "True", "descricao": "Atributo TP_LEVANTA preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "305", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_NR_ESCALAO", "incluida_no_csv_leve": "True", "descricao": "Atributo NR_ESCALAO preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "301", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_AREA_CALC", "incluida_no_csv_leve": "True", "descricao": "Atributo AREA_CALC preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "35", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_PERIMETRO", "incluida_no_csv_leve": "True", "descricao": "Atributo PERIMETRO preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "35", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_ESFERA", "incluida_no_csv_leve": "True", "descricao": "Atributo de esfera selecionado como dimensão do relatório de quilombolas.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "34", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1swdqtJoEeWwDhmpmYkPAzP231q6FjBj3/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_FASE", "incluida_no_csv_leve": "True", "descricao": "Fase do território quilombola selecionada no relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "34", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1swdqtJoEeWwDhmpmYkPAzP231q6FjBj3/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_RESPONSAVE", "incluida_no_csv_leve": "True", "descricao": "Atributo RESPONSAVE preservado da camada de origem com prefixo QUIL. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "35", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Identificador sequencial de feição atribuído no notebook de interseção, iniciado em 1 após a preparação da camada. Depende da versão e ordem da base; não é identificador oficial do órgão.", "tipo_logico": "identificador interno", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "atribuição np.arange(1, len(camada) + 1)", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "QUIL_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Área da feição completa da camada, calculada no notebook de interseção após projeção EPSG:5880: geometry.area / 10000. Não é área de interseção nem área documental declarada.", "tipo_logico": "numérico de origem", "unidade": "ha", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1LtySu-_CtPZMz4vr1DY76q78AteoyRnK/view", "criterio_revisao": "CRS_AREA e cálculo da área da camada", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do programa de crédito utilizado no vínculo com o arquivo Programa.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do programa de crédito obtida por junção com a tabela de programas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Descrição da categoria do emitente obtida por junção com a tabela de categorias.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Classificação analítica por ordem: pequeno/Pronaf, médio/Pronamp, grande ou demais/não classificado. Base04 usa, nessa precedência, categoria do emitente e descrições de categoria e programa; não constitui nova classificação oficial do beneficiário.", "tipo_logico": "categoria analítica", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "quilombolas", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ANO_OPERACAO no processamento anterior; não é uma nova observação independente. Em quilombolas, a origem indicada é ANO_OPERACAO. Não equiparar automaticamente ano do arquivo e ano de emissão.", "tipo_logico": "ano", "unidade": "ano", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1swdqtJoEeWwDhmpmYkPAzP231q6FjBj3/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ANO_OPERACAO", "teste_derivacao": "numero"}, {"conjunto": "quilombolas", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de QUIL_FASE no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1swdqtJoEeWwDhmpmYkPAzP231q6FjBj3/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "QUIL_FASE", "teste_derivacao": "texto"}, {"conjunto": "quilombolas", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TIPO_IMPEDIMENTO no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1swdqtJoEeWwDhmpmYkPAzP231q6FjBj3/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "A coluna selecionada é TIPO_IMPEDIMENTO, rótulo da camada; não comprova modalidade específica do território.", "coluna_derivadora": "TIPO_IMPEDIMENTO", "teste_derivacao": "texto"}, {"conjunto": "quilombolas", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de QUIL_ESFERA no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1swdqtJoEeWwDhmpmYkPAzP231q6FjBj3/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "QUIL_ESFERA", "teste_derivacao": "texto"}, {"conjunto": "quilombolas", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de QUIL_CD_UF no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "368", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1swdqtJoEeWwDhmpmYkPAzP231q6FjBj3/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "QUIL_CD_UF", "teste_derivacao": "texto"}, {"conjunto": "florestas_publicas", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Índice da geometria dentro da ordem, preservado da base territorial.", "tipo_logico": "identificador", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Código da categoria do emitente utilizado no vínculo com o arquivo de categorias.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Campo de estado preservado da base operacional. O código Base04 tenta utilizá-lo no vínculo com siglas de UF; não é uma UF calculada pela interseção.", "tipo_logico": "código de origem", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do programa de crédito utilizado no vínculo com o arquivo Programa.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Área financiada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "3", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do subprograma de crédito utilizado no vínculo com o arquivo Subprogramas.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "8266", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Área informada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "3", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do programa de crédito obtida por junção com a tabela de programas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Descrição da categoria do emitente obtida por junção com a tabela de categorias.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do subprograma obtida por junção com a tabela de subprogramas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "8266", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Quantidade de linhas da base operacional agrupadas pela chave REF_BACEN e ordem. Não é número de glebas.", "tipo_logico": "inteiro", "unidade": "linhas", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Classificação analítica por ordem: pequeno/Pronaf, médio/Pronamp, grande ou demais/não classificado. Base04 usa, nessa precedência, categoria do emitente e descrições de categoria e programa; não constitui nova classificação oficial do beneficiário.", "tipo_logico": "categoria analítica", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Sim quando a descrição do subprograma contém GRUPO B (expressão GRUPO\\s*B); Não nos demais casos. É identificação textual realizada por Base04.", "tipo_logico": "indicador textual", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Vinculado quando REF_BACEN e ordem são encontrados na junção com a base operacional; Não vinculado quando não encontrados.", "tipo_logico": "categoria de vínculo", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_NOME", "incluida_no_csv_leve": "True", "descricao": "Nome da floresta pública utilizado pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/17XKmLDTVU4CbTQx3hyHQY3QrJsAQdjQQ/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_CATEGORIA", "incluida_no_csv_leve": "True", "descricao": "Atributo CATEGORIA preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_PROTECAO", "incluida_no_csv_leve": "True", "descricao": "Atributo PROTECAO preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_TIPO", "incluida_no_csv_leve": "True", "descricao": "Tipo de floresta pública utilizado pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/17XKmLDTVU4CbTQx3hyHQY3QrJsAQdjQQ/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_GOVERNO", "incluida_no_csv_leve": "True", "descricao": "Atributo GOVERNO preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_COMUNITARI", "incluida_no_csv_leve": "True", "descricao": "Atributo COMUNITARI preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_CLASSE", "incluida_no_csv_leve": "True", "descricao": "Atributo CLASSE preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_ORGAO", "incluida_no_csv_leve": "True", "descricao": "Atributo ORGAO preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1016", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_SOBREPOSIC", "incluida_no_csv_leve": "True", "descricao": "Atributo SOBREPOSIC preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_ATOLEGAL", "incluida_no_csv_leve": "True", "descricao": "Atributo ATOLEGAL preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "22994", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_ANOCRIACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo ANOCRIACAO preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "3412", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_ESTAGIO", "incluida_no_csv_leve": "True", "descricao": "Atributo ESTAGIO preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "17901", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_CODIGO", "incluida_no_csv_leve": "True", "descricao": "Atributo CODIGO preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "33984", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_OBSERVACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo OBSERVACAO preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "33984", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_UF", "incluida_no_csv_leve": "True", "descricao": "UF informada na camada de florestas públicas utilizada pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/17XKmLDTVU4CbTQx3hyHQY3QrJsAQdjQQ/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_SHAPE_LENG", "incluida_no_csv_leve": "True", "descricao": "Atributo SHAPE_LENG preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_SHAPE_AREA", "incluida_no_csv_leve": "True", "descricao": "Atributo SHAPE_AREA preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_BIOMA", "incluida_no_csv_leve": "True", "descricao": "Atributo BIOMA preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo MUNICIPIO preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_GEOCODIGO", "incluida_no_csv_leve": "True", "descricao": "Atributo GEOCODIGO preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_AREA_HA", "incluida_no_csv_leve": "True", "descricao": "Atributo AREA_HA preservado da camada de origem com prefixo FP. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Identificador sequencial de feição atribuído no notebook de interseção, iniciado em 1 após a preparação da camada. Depende da versão e ordem da base; não é identificador oficial do órgão.", "tipo_logico": "identificador interno", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "atribuição np.arange(1, len(camada) + 1)", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "FP_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Área da feição completa da camada, calculada no notebook de interseção após projeção EPSG:5880: geometry.area / 10000. Não é área de interseção nem área documental declarada.", "tipo_logico": "numérico de origem", "unidade": "ha", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_/view", "criterio_revisao": "CRS_AREA e cálculo da área da camada", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "florestas_publicas", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ANO_BASE no processamento anterior; não é uma nova observação independente. Em florestas_publicas, a origem indicada é ANO_BASE. Não equiparar automaticamente ano do arquivo e ano de emissão.", "tipo_logico": "ano", "unidade": "ano", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/17XKmLDTVU4CbTQx3hyHQY3QrJsAQdjQQ/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ANO_BASE", "teste_derivacao": "numero"}, {"conjunto": "florestas_publicas", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de FP_TIPO no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/17XKmLDTVU4CbTQx3hyHQY3QrJsAQdjQQ/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "FP_TIPO", "teste_derivacao": "texto"}, {"conjunto": "florestas_publicas", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar sem coluna-fonte selecionada: o código preenche Não informado.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/17XKmLDTVU4CbTQx3hyHQY3QrJsAQdjQQ/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": "constante"}, {"conjunto": "florestas_publicas", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar sem coluna-fonte selecionada: o código preenche Não informado.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/17XKmLDTVU4CbTQx3hyHQY3QrJsAQdjQQ/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": "constante"}, {"conjunto": "florestas_publicas", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de FP_UF no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33984", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/17XKmLDTVU4CbTQx3hyHQY3QrJsAQdjQQ/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "FP_UF", "teste_derivacao": "texto"}, {"conjunto": "assentamentos", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Índice da geometria dentro da ordem, preservado da base territorial.", "tipo_logico": "identificador", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Código da categoria do emitente utilizado no vínculo com o arquivo de categorias.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Campo de estado preservado da base operacional. O código Base04 tenta utilizá-lo no vínculo com siglas de UF; não é uma UF calculada pela interseção.", "tipo_logico": "código de origem", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do programa de crédito utilizado no vínculo com o arquivo Programa.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Área financiada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do subprograma de crédito utilizado no vínculo com o arquivo Subprogramas.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "3756", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Área informada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do programa de crédito obtida por junção com a tabela de programas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Descrição da categoria do emitente obtida por junção com a tabela de categorias.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do subprograma obtida por junção com a tabela de subprogramas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "3756", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Quantidade de linhas da base operacional agrupadas pela chave REF_BACEN e ordem. Não é número de glebas.", "tipo_logico": "inteiro", "unidade": "linhas", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Classificação analítica por ordem: pequeno/Pronaf, médio/Pronamp, grande ou demais/não classificado. Base04 usa, nessa precedência, categoria do emitente e descrições de categoria e programa; não constitui nova classificação oficial do beneficiário.", "tipo_logico": "categoria analítica", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Sim quando a descrição do subprograma contém GRUPO B (expressão GRUPO\\s*B); Não nos demais casos. É identificação textual realizada por Base04.", "tipo_logico": "indicador textual", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Vinculado quando REF_BACEN e ordem são encontrados na junção com a base operacional; Não vinculado quando não encontrados.", "tipo_logico": "categoria de vínculo", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_CD_SIPRA", "incluida_no_csv_leve": "True", "descricao": "Atributo CD_SIPRA preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo UF preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_NOME_PROJE", "incluida_no_csv_leve": "True", "descricao": "Atributo NOME_PROJE preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo MUNICIPIO preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_AREA_HECTA", "incluida_no_csv_leve": "True", "descricao": "Atributo AREA_HECTA preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_CAPACIDADE", "incluida_no_csv_leve": "True", "descricao": "Atributo CAPACIDADE preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_NUM_FAMILI", "incluida_no_csv_leve": "True", "descricao": "Atributo NUM_FAMILI preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_FASE", "incluida_no_csv_leve": "True", "descricao": "Atributo FASE preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_DATA_DE_CR", "incluida_no_csv_leve": "True", "descricao": "Atributo DATA_DE_CR preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_FORMA_OBTE", "incluida_no_csv_leve": "True", "descricao": "Atributo FORMA_OBTE preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "20", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_DATA_OBTEN", "incluida_no_csv_leve": "True", "descricao": "Atributo DATA_OBTEN preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "20", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_AREA_CALC", "incluida_no_csv_leve": "True", "descricao": "Atributo AREA_CALC preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1188", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_SR", "incluida_no_csv_leve": "True", "descricao": "Atributo SR preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "33044", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_DESCRICAO", "incluida_no_csv_leve": "True", "descricao": "Atributo DESCRICAO preservado da camada de origem com prefixo ASSENT. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "33044", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Identificador sequencial de feição atribuído no notebook de interseção, iniciado em 1 após a preparação da camada. Depende da versão e ordem da base; não é identificador oficial do órgão.", "tipo_logico": "identificador interno", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "atribuição np.arange(1, len(camada) + 1)", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "ASSENT_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Área da feição completa da camada, calculada no notebook de interseção após projeção EPSG:5880: geometry.area / 10000. Não é área de interseção nem área documental declarada.", "tipo_logico": "numérico de origem", "unidade": "ha", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U/view", "criterio_revisao": "CRS_AREA e cálculo da área da camada", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "assentamentos", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ANO_BASE no processamento anterior; não é uma nova observação independente. Em assentamentos, a origem indicada é ANO_BASE. Não equiparar automaticamente ano do arquivo e ano de emissão.", "tipo_logico": "ano", "unidade": "ano", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bz4lsYYYPLF1eB8rrNZc99YxzJe0UuhE/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ANO_BASE", "teste_derivacao": "numero"}, {"conjunto": "assentamentos", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ASSENT_FASE no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bz4lsYYYPLF1eB8rrNZc99YxzJe0UuhE/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ASSENT_FASE", "teste_derivacao": "texto"}, {"conjunto": "assentamentos", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TIPO_PRODUTOR_ANALISE no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bz4lsYYYPLF1eB8rrNZc99YxzJe0UuhE/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "A busca por substring no código pode selecionar a classificação do produtor no lugar do tipo do assentamento. Confirmar pela comparação automática com o CSV leve.", "coluna_derivadora": "TIPO_PRODUTOR_ANALISE", "teste_derivacao": "texto"}, {"conjunto": "assentamentos", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ASSENT_FORMA_OBTE no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bz4lsYYYPLF1eB8rrNZc99YxzJe0UuhE/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ASSENT_FORMA_OBTE", "teste_derivacao": "texto"}, {"conjunto": "assentamentos", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ASSENT_UF no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "33044", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1bz4lsYYYPLF1eB8rrNZc99YxzJe0UuhE/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ASSENT_UF", "teste_derivacao": "texto"}, {"conjunto": "embargos", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Índice da geometria dentro da ordem, preservado da base territorial.", "tipo_logico": "identificador", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Código da categoria do emitente utilizado no vínculo com o arquivo de categorias.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Campo de estado preservado da base operacional. O código Base04 tenta utilizá-lo no vínculo com siglas de UF; não é uma UF calculada pela interseção.", "tipo_logico": "código de origem", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do programa de crédito utilizado no vínculo com o arquivo Programa.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Área financiada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do subprograma de crédito utilizado no vínculo com o arquivo Subprogramas.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "723", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Área informada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do programa de crédito obtida por junção com a tabela de programas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Descrição da categoria do emitente obtida por junção com a tabela de categorias.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do subprograma obtida por junção com a tabela de subprogramas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "723", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Quantidade de linhas da base operacional agrupadas pela chave REF_BACEN e ordem. Não é número de glebas.", "tipo_logico": "inteiro", "unidade": "linhas", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Classificação analítica por ordem: pequeno/Pronaf, médio/Pronamp, grande ou demais/não classificado. Base04 usa, nessa precedência, categoria do emitente e descrições de categoria e programa; não constitui nova classificação oficial do beneficiário.", "tipo_logico": "categoria analítica", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Sim quando a descrição do subprograma contém GRUPO B (expressão GRUPO\\s*B); Não nos demais casos. É identificação textual realizada por Base04.", "tipo_logico": "indicador textual", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Vinculado quando REF_BACEN e ordem são encontrados na junção com a base operacional; Não vinculado quando não encontrados.", "tipo_logico": "categoria de vínculo", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_OBJECTID", "incluida_no_csv_leve": "True", "descricao": "Atributo OBJECTID preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_SEQ_TAD", "incluida_no_csv_leve": "True", "descricao": "Atributo SEQ_TAD preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_NUM_TAD", "incluida_no_csv_leve": "True", "descricao": "Atributo NUM_TAD preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_SERIE_TAD", "incluida_no_csv_leve": "True", "descricao": "Atributo SERIE_TAD preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "567", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo OPERACAO preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "884", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_ORIGEM_GEO", "incluida_no_csv_leve": "True", "descricao": "Atributo ORIGEM_GEO preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_COD_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo COD_UF preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo UF preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_COD_MUNICI", "incluida_no_csv_leve": "True", "descricao": "Atributo COD_MUNICI preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo MUNICIPIO preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_NOME_IMOVE", "incluida_no_csv_leve": "True", "descricao": "Nome do imóvel selecionado como nome da unidade no relatório de embargos.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "753", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1X3KBwA9LXNmqHFEYlvNXsgHpEQVOoYU8/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_DES_LOCALI", "incluida_no_csv_leve": "True", "descricao": "Atributo DES_LOCALI preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "180", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_NOME_EMBAR", "incluida_no_csv_leve": "True", "descricao": "Atributo NOME_EMBAR preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "25", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_CPF_CNPJ_E", "incluida_no_csv_leve": "True", "descricao": "Atributo CPF_CNPJ_E preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "26", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_SIT_DESMAT", "incluida_no_csv_leve": "True", "descricao": "Atributo SIT_DESMAT preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_TIPO_AREA", "incluida_no_csv_leve": "True", "descricao": "Atributo TIPO_AREA preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "279", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_NUM_AUTO_I", "incluida_no_csv_leve": "True", "descricao": "Atributo NUM_AUTO_I preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "133", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_SERIE_AUTO", "incluida_no_csv_leve": "True", "descricao": "Atributo SERIE_AUTO preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "594", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_COD_TIPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo COD_TIPO_B preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "303", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_DES_TIPO_B", "incluida_no_csv_leve": "True", "descricao": "Atributo DES_TIPO_B preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "303", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_UNID_CONTR", "incluida_no_csv_leve": "True", "descricao": "Atributo UNID_CONTR preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_ORDEM_FISC", "incluida_no_csv_leve": "True", "descricao": "Atributo ORDEM_FISC preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "278", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_CD_ACAO_FI", "incluida_no_csv_leve": "True", "descricao": "Atributo CD_ACAO_FI preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "405", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_NUM_PROCES", "incluida_no_csv_leve": "True", "descricao": "Atributo NUM_PROCES preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "46", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_DES_TAD", "incluida_no_csv_leve": "True", "descricao": "Atributo DES_TAD preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_DES_INFRAC", "incluida_no_csv_leve": "True", "descricao": "Atributo DES_INFRAC preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "215", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_NUM_LONGIT", "incluida_no_csv_leve": "True", "descricao": "Atributo NUM_LONGIT preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_NUM_LATITU", "incluida_no_csv_leve": "True", "descricao": "Atributo NUM_LATITU preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "13", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_DAT_EMBARG", "incluida_no_csv_leve": "True", "descricao": "Atributo DAT_EMBARG preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_DAT_IMPRES", "incluida_no_csv_leve": "True", "descricao": "Atributo DAT_IMPRES preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "53", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_DAT_ULT_AL", "incluida_no_csv_leve": "True", "descricao": "Atributo DAT_ULT_AL preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_NUM_LONG00", "incluida_no_csv_leve": "True", "descricao": "Atributo NUM_LONG00 preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_NUM_LATI00", "incluida_no_csv_leve": "True", "descricao": "Atributo NUM_LATI00 preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "13", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_QTD_AREA_D", "incluida_no_csv_leve": "True", "descricao": "Atributo QTD_AREA_D preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "648", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_QTD_AREA_E", "incluida_no_csv_leve": "True", "descricao": "Atributo QTD_AREA_E preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "56", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_DAT_ULT_00", "incluida_no_csv_leve": "True", "descricao": "Atributo DAT_ULT_00 preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_ST_AREA_SH", "incluida_no_csv_leve": "True", "descricao": "Atributo ST_AREA_SH preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_ST_PERIMET", "incluida_no_csv_leve": "True", "descricao": "Atributo ST_PERIMET preservado da camada de origem com prefixo EMB. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Identificador sequencial de feição atribuído no notebook de interseção, iniciado em 1 após a preparação da camada. Depende da versão e ordem da base; não é identificador oficial do órgão.", "tipo_logico": "identificador interno", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "atribuição np.arange(1, len(camada) + 1)", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "EMB_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Área da feição completa da camada, calculada no notebook de interseção após projeção EPSG:5880: geometry.area / 10000. Não é área de interseção nem área documental declarada.", "tipo_logico": "numérico de origem", "unidade": "ha", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7/view", "criterio_revisao": "CRS_AREA e cálculo da área da camada", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "embargos", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ANO_BASE no processamento anterior; não é uma nova observação independente. Em embargos, a origem indicada é ANO_BASE. Não equiparar automaticamente ano do arquivo e ano de emissão.", "tipo_logico": "ano", "unidade": "ano", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1X3KBwA9LXNmqHFEYlvNXsgHpEQVOoYU8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ANO_BASE", "teste_derivacao": "numero"}, {"conjunto": "embargos", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar sem coluna-fonte selecionada: o código preenche Não informado.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1X3KBwA9LXNmqHFEYlvNXsgHpEQVOoYU8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": "constante"}, {"conjunto": "embargos", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar sem coluna-fonte selecionada: o código preenche Não informado.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1X3KBwA9LXNmqHFEYlvNXsgHpEQVOoYU8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": "constante"}, {"conjunto": "embargos", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TIPO_PRODUTOR_ANALISE no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1X3KBwA9LXNmqHFEYlvNXsgHpEQVOoYU8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "O rótulo do relatório é tipo de infração/embargo, mas a coluna selecionada contém classificação do produtor. Não usar como tipo de infração.", "coluna_derivadora": "TIPO_PRODUTOR_ANALISE", "teste_derivacao": "texto"}, {"conjunto": "embargos", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de UF_OPERACAO no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "931", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1X3KBwA9LXNmqHFEYlvNXsgHpEQVOoYU8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "Campo derivado de UF_OPERACAO; não representa localização espacial do embargo.", "coluna_derivadora": "UF_OPERACAO", "teste_derivacao": "texto"}, {"conjunto": "unidades_conservacao", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Índice da geometria dentro da ordem, preservado da base territorial.", "tipo_logico": "identificador", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Código da categoria do emitente utilizado no vínculo com o arquivo de categorias.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Campo de estado preservado da base operacional. O código Base04 tenta utilizá-lo no vínculo com siglas de UF; não é uma UF calculada pela interseção.", "tipo_logico": "código de origem", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do programa de crédito utilizado no vínculo com o arquivo Programa.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Área financiada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "5", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do subprograma de crédito utilizado no vínculo com o arquivo Subprogramas.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "10074", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Área informada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "5", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do programa de crédito obtida por junção com a tabela de programas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Descrição da categoria do emitente obtida por junção com a tabela de categorias.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do subprograma obtida por junção com a tabela de subprogramas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "10074", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Quantidade de linhas da base operacional agrupadas pela chave REF_BACEN e ordem. Não é número de glebas.", "tipo_logico": "inteiro", "unidade": "linhas", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Classificação analítica por ordem: pequeno/Pronaf, médio/Pronamp, grande ou demais/não classificado. Base04 usa, nessa precedência, categoria do emitente e descrições de categoria e programa; não constitui nova classificação oficial do beneficiário.", "tipo_logico": "categoria analítica", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Sim quando a descrição do subprograma contém GRUPO B (expressão GRUPO\\s*B); Não nos demais casos. É identificação textual realizada por Base04.", "tipo_logico": "indicador textual", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Vinculado quando REF_BACEN e ordem são encontrados na junção com a base operacional; Não vinculado quando não encontrados.", "tipo_logico": "categoria de vínculo", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_GML_ID", "incluida_no_csv_leve": "True", "descricao": "Atributo GML_ID preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18617", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_DOCLEG_ID", "incluida_no_csv_leve": "True", "descricao": "Atributo DOCLEG_ID preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_UC_ID", "incluida_no_csv_leve": "True", "descricao": "Atributo UC_ID preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_CD_CNUC", "incluida_no_csv_leve": "True", "descricao": "Atributo CD_CNUC preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_WDPA_PID", "incluida_no_csv_leve": "True", "descricao": "Atributo WDPA_PID preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "1667", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_NOME_UC", "incluida_no_csv_leve": "True", "descricao": "Nome da unidade de conservação utilizado pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_CRIA_ANO", "incluida_no_csv_leve": "True", "descricao": "Atributo CRIA_ANO preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_CRIA_ATO", "incluida_no_csv_leve": "True", "descricao": "Atributo CRIA_ATO preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_OUTRO_ATO", "incluida_no_csv_leve": "True", "descricao": "Atributo OUTRO_ATO preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_PL_MANEJO", "incluida_no_csv_leve": "True", "descricao": "Atributo PL_MANEJO preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_CO_GESTOR", "incluida_no_csv_leve": "True", "descricao": "Atributo CO_GESTOR preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_QUALI_POL", "incluida_no_csv_leve": "True", "descricao": "Atributo QUALI_POL preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_PPGR", "incluida_no_csv_leve": "True", "descricao": "Atributo PPGR preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "14893", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_HA_TOTAL", "incluida_no_csv_leve": "True", "descricao": "Atributo HA_TOTAL preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "40", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_HA_ATO", "incluida_no_csv_leve": "True", "descricao": "Atributo HA_ATO preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_ESFERA", "incluida_no_csv_leve": "True", "descricao": "Esfera administrativa da unidade de conservação utilizada pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_UF", "incluida_no_csv_leve": "True", "descricao": "UF informada na camada de unidades de conservação utilizada pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo MUNICIPIO preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_ORG_GESTOR", "incluida_no_csv_leve": "True", "descricao": "Atributo ORG_GESTOR preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_GRUPO", "incluida_no_csv_leve": "True", "descricao": "Grupo de proteção da unidade de conservação utilizado pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_CATEGORIA", "incluida_no_csv_leve": "True", "descricao": "Categoria de manejo da unidade de conservação utilizada pelo relatório.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_CAT_IUCN", "incluida_no_csv_leve": "True", "descricao": "Atributo CAT_IUCN preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_AMAZONIA", "incluida_no_csv_leve": "True", "descricao": "Atributo AMAZONIA preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18011", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_CAATINGA", "incluida_no_csv_leve": "True", "descricao": "Atributo CAATINGA preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "15324", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_CERRADO", "incluida_no_csv_leve": "True", "descricao": "Atributo CERRADO preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "40", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_MATLANTICA", "incluida_no_csv_leve": "True", "descricao": "Atributo MATLANTICA preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "15535", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_PAMPA", "incluida_no_csv_leve": "True", "descricao": "Atributo PAMPA preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18617", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_PANTANAL", "incluida_no_csv_leve": "True", "descricao": "Atributo PANTANAL preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "18616", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_MARINHO", "incluida_no_csv_leve": "True", "descricao": "Atributo MARINHO preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "17664", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_SITUACAO", "incluida_no_csv_leve": "True", "descricao": "Atributo SITUACAO preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_LIMITE", "incluida_no_csv_leve": "True", "descricao": "Atributo LIMITE preservado da camada de origem com prefixo UC. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Identificador sequencial de feição atribuído no notebook de interseção, iniciado em 1 após a preparação da camada. Depende da versão e ordem da base; não é identificador oficial do órgão.", "tipo_logico": "identificador interno", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "atribuição np.arange(1, len(camada) + 1)", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "UC_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Área da feição completa da camada, calculada no notebook de interseção após projeção EPSG:5880: geometry.area / 10000. Não é área de interseção nem área documental declarada.", "tipo_logico": "numérico de origem", "unidade": "ha", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj/view", "criterio_revisao": "CRS_AREA e cálculo da área da camada", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "unidades_conservacao", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ANO_BASE no processamento anterior; não é uma nova observação independente. Em unidades_conservacao, a origem indicada é ANO_BASE. Não equiparar automaticamente ano do arquivo e ano de emissão.", "tipo_logico": "ano", "unidade": "ano", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ANO_BASE", "teste_derivacao": "numero"}, {"conjunto": "unidades_conservacao", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de UC_CATEGORIA no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "UC_CATEGORIA", "teste_derivacao": "texto"}, {"conjunto": "unidades_conservacao", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de UC_GRUPO no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "UC_GRUPO", "teste_derivacao": "texto"}, {"conjunto": "unidades_conservacao", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de UC_ESFERA no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "UC_ESFERA", "teste_derivacao": "texto"}, {"conjunto": "unidades_conservacao", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de UC_UF no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "18617", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "UC_UF", "teste_derivacao": "texto"}, {"conjunto": "car_suspenso", "coluna": "NU_INDICE", "incluida_no_csv_leve": "True", "descricao": "Índice da geometria dentro da ordem, preservado da base territorial.", "tipo_logico": "identificador", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CD_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Código da categoria do emitente utilizado no vínculo com o arquivo de categorias.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CD_ESTADO", "incluida_no_csv_leve": "True", "descricao": "Campo de estado preservado da base operacional. O código Base04 tenta utilizá-lo no vínculo com siglas de UF; não é uma UF calculada pela interseção.", "tipo_logico": "código de origem", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CD_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do programa de crédito utilizado no vínculo com o arquivo Programa.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "VL_AREA_FINANC", "incluida_no_csv_leve": "True", "descricao": "Área financiada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CD_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Código do subprograma de crédito utilizado no vínculo com o arquivo Subprogramas.csv.", "tipo_logico": "código", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "193", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "VL_AREA_INFORMADA", "incluida_no_csv_leve": "True", "descricao": "Área informada preservada da base operacional; Base04 reconhece o campo, mas não documenta sua unidade. Não equiparar à área calculada da geometria.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "DESC_PROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do programa de crédito obtida por junção com a tabela de programas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "DESC_CATEG_EMITENTE", "incluida_no_csv_leve": "True", "descricao": "Descrição da categoria do emitente obtida por junção com a tabela de categorias.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "DESC_SUBPROGRAMA", "incluida_no_csv_leve": "True", "descricao": "Descrição do subprograma obtida por junção com a tabela de subprogramas.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "193", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "QUANTIDADE_LINHAS_OPERACIONAIS_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Quantidade de linhas da base operacional agrupadas pela chave REF_BACEN e ordem. Não é número de glebas.", "tipo_logico": "inteiro", "unidade": "linhas", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "TIPO_PRODUTOR_ANALISE", "incluida_no_csv_leve": "True", "descricao": "Classificação analítica por ordem: pequeno/Pronaf, médio/Pronamp, grande ou demais/não classificado. Base04 usa, nessa precedência, categoria do emitente e descrições de categoria e programa; não constitui nova classificação oficial do beneficiário.", "tipo_logico": "categoria analítica", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "IND_PRONAF_GRUPO_B", "incluida_no_csv_leve": "True", "descricao": "Sim quando a descrição do subprograma contém GRUPO B (expressão GRUPO\\s*B); Não nos demais casos. É identificação textual realizada por Base04.", "tipo_logico": "indicador textual", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "VINCULO_OPERACAO", "incluida_no_csv_leve": "True", "descricao": "Vinculado quando REF_BACEN e ordem são encontrados na junção com a base operacional; Não vinculado quando não encontrados.", "tipo_logico": "categoria de vínculo", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1auwGog0NeTC5T_rO25ZsdAu55vF9rAqN/view", "criterio_revisao": "aliases, junções, classificações e dicionário definidos em Base04", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_COD_IMOVEL", "incluida_no_csv_leve": "True", "descricao": "Código do cadastro imobiliário do CAR selecionado como identificador analítico da unidade. Não é prova de domínio.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1HNYfKtqvlLtgc1lj60s1kV04KOoF_Jyi/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_STATUS_IMO", "incluida_no_csv_leve": "True", "descricao": "Situação cadastral selecionada no relatório do CAR. Representa a base utilizada, não a situação histórica na contratação.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1HNYfKtqvlLtgc1lj60s1kV04KOoF_Jyi/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_DAT_CRIACA", "incluida_no_csv_leve": "True", "descricao": "Atributo DAT_CRIACA preservado da camada de origem com prefixo CAR. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_DATA_ATUAL", "incluida_no_csv_leve": "True", "descricao": "Atributo DATA_ATUAL preservado da camada de origem com prefixo CAR. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "34", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_AREA", "incluida_no_csv_leve": "True", "descricao": "Atributo AREA preservado da camada de origem com prefixo CAR. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_CONDICAO", "incluida_no_csv_leve": "True", "descricao": "Condição cadastral complementar selecionada no relatório do CAR.", "tipo_logico": "texto/código de origem", "unidade": "não aplicável", "situacao_definicao": "definição de uso documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1HNYfKtqvlLtgc1lj60s1kV04KOoF_Jyi/view", "criterio_revisao": "seleção explícita ou mapeamento registrado no relatório de origem; domínio completo de códigos não revisado", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_UF", "incluida_no_csv_leve": "True", "descricao": "Atributo UF preservado da camada de origem com prefixo CAR. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_MUNICIPIO", "incluida_no_csv_leve": "True", "descricao": "Atributo MUNICIPIO preservado da camada de origem com prefixo CAR. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_COD_MUNICI", "incluida_no_csv_leve": "True", "descricao": "Atributo COD_MUNICI preservado da camada de origem com prefixo CAR. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_M_FISCAL", "incluida_no_csv_leve": "True", "descricao": "Atributo M_FISCAL preservado da camada de origem com prefixo CAR. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_TIPO_IMOVE", "incluida_no_csv_leve": "True", "descricao": "Atributo TIPO_IMOVE preservado da camada de origem com prefixo CAR. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_STATUS_CAR_CODIGO", "incluida_no_csv_leve": "True", "descricao": "Atributo STATUS_CAR_CODIGO preservado da camada de origem com prefixo CAR. O código comprova sua cópia, mas não documenta integralmente significado, domínio e unidade.", "tipo_logico": "texto de origem", "unidade": "não confirmada", "situacao_definicao": "definição pendente", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "proveniência identificada; definição semântica não confirmada", "observacao_revisao": "Manter pendente. Não inferir a definição apenas pela abreviação; consultar metadados/dicionário da fonte.", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_UF_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "UF do arquivo estadual lido no laço de processamento; não é resultado de interseção com limites estaduais.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "atribuições UF_ORIGEM e ARQUIVO_ORIGEM antes da concatenação", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_ARQUIVO_ORIGEM", "incluida_no_csv_leve": "True", "descricao": "Nome do arquivo estadual de CAR lido no processamento.", "tipo_logico": "texto", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "atribuições UF_ORIGEM e ARQUIVO_ORIGEM antes da concatenação", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_ID_INTERNO", "incluida_no_csv_leve": "True", "descricao": "Identificador sequencial de feição atribuído no notebook de interseção, iniciado em 1 após a preparação da camada. Depende da versão e ordem da base; não é identificador oficial do órgão.", "tipo_logico": "identificador interno", "unidade": "não aplicável", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "atribuição np.arange(1, len(camada) + 1)", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "CAR_AREA_TOTAL_HA", "incluida_no_csv_leve": "True", "descricao": "Área da feição completa da camada, calculada no notebook de interseção após projeção EPSG:5880: geometry.area / 10000. Não é área de interseção nem área documental declarada.", "tipo_logico": "numérico de origem", "unidade": "ha", "situacao_definicao": "definição documentada no código", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK/view", "criterio_revisao": "CRS_AREA e cálculo da área da camada", "observacao_revisao": "", "coluna_derivadora": "", "teste_derivacao": ""}, {"conjunto": "car_suspenso", "coluna": "_ANO", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de ANO_BASE no processamento anterior; não é uma nova observação independente. Em car_suspenso, a origem indicada é ANO_BASE. Não equiparar automaticamente ano do arquivo e ano de emissão.", "tipo_logico": "ano", "unidade": "ano", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1HNYfKtqvlLtgc1lj60s1kV04KOoF_Jyi/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "ANO_BASE", "teste_derivacao": "numero"}, {"conjunto": "car_suspenso", "coluna": "_TIPOLOGIA_1", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de CAR_STATUS_IMO no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1HNYfKtqvlLtgc1lj60s1kV04KOoF_Jyi/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "CAR_STATUS_IMO", "teste_derivacao": "texto"}, {"conjunto": "car_suspenso", "coluna": "_TIPOLOGIA_2", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de TIPO_PRODUTOR_ANALISE no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1HNYfKtqvlLtgc1lj60s1kV04KOoF_Jyi/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "O rótulo do relatório é tipo de imóvel, mas a coluna selecionada contém classificação do produtor. Não usar como tipo de imóvel.", "coluna_derivadora": "TIPO_PRODUTOR_ANALISE", "teste_derivacao": "texto"}, {"conjunto": "car_suspenso", "coluna": "_TIPOLOGIA_3", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de CAR_CONDICAO no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1HNYfKtqvlLtgc1lj60s1kV04KOoF_Jyi/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "", "coluna_derivadora": "CAR_CONDICAO", "teste_derivacao": "texto"}, {"conjunto": "car_suspenso", "coluna": "_UF_UNIDADE", "incluida_no_csv_leve": "True", "descricao": "Campo auxiliar derivado de UF_OPERACAO no processamento anterior; não é uma nova observação independente.", "tipo_logico": "atributo derivado", "unidade": "não aplicável", "situacao_definicao": "definição documentada; conferir derivação no CSV", "vazios": "0", "registros": "530", "armazenamento": "texto; valores e separadores decimais preservados", "fonte_definicao": "https://drive.google.com/file/d/1HNYfKtqvlLtgc1lj60s1kV04KOoF_Jyi/view", "criterio_revisao": "atribuições, seleção de colunas e saídas dos notebooks; confirmar no CSV leve", "observacao_revisao": "Campo derivado de UF_OPERACAO; não representa localização espacial do CAR.", "coluna_derivadora": "UF_OPERACAO", "teste_derivacao": "texto"}], "fontes_notebooks": {"terras_indigenas": "https://drive.google.com/file/d/1LY7W-oRvdpaMYRx7FIAfDLulhSZW7ESr/view", "quilombolas": "https://drive.google.com/file/d/1swdqtJoEeWwDhmpmYkPAzP231q6FjBj3/view", "florestas_publicas": "https://drive.google.com/file/d/17XKmLDTVU4CbTQx3hyHQY3QrJsAQdjQQ/view", "assentamentos": "https://drive.google.com/file/d/1bz4lsYYYPLF1eB8rrNZc99YxzJe0UuhE/view", "embargos": "https://drive.google.com/file/d/1X3KBwA9LXNmqHFEYlvNXsgHpEQVOoYU8/view", "unidades_conservacao": "https://drive.google.com/file/d/1isujDEJjbttMNNRU0kUMcxe2QQkq-dD8/view", "car_suspenso": "https://drive.google.com/file/d/1HNYfKtqvlLtgc1lj60s1kV04KOoF_Jyi/view"}, "fontes_intersecoes": {"terras_indigenas": "1bJVm_QXILKwsCjRWoFF1VEo2F7Zoo6ji", "quilombolas": "1LtySu-_CtPZMz4vr1DY76q78AteoyRnK", "florestas_publicas": "1qBAIgx1OhXBFhvLafrfowe9hVulUZIr_", "assentamentos": "1R03vVziqGhEaGdtg5qEZ1BEbtpi7-Z9U", "embargos": "15Qy-Nxe1hnEY5YX4myzaS_aRFGR2qtq7", "unidades_conservacao": "1b7zPOS0MXwGkAEMSzEsbhMpwNXTMTcvj", "car_suspenso": "1oWoVzLsmy5XpWSpamH4ZD5FBd2CEiUUK"}}''')
from pathlib import Path
from datetime import datetime, timezone
import json, hashlib, shutil, zipfile
import pandas as pd

def hash_arquivo(p):
    h=hashlib.sha256()
    with Path(p).open('rb') as f:
        for bloco in iter(lambda:f.read(1024*1024),b''):h.update(bloco)
    return h.hexdigest()

def ler_csv(p):
    return pd.read_csv(p,sep=';',dtype=str,keep_default_na=False)

def texto_limpo(s):
    return s.astype(str).str.strip().str.replace(r'\.0$','',regex=True).replace({'':'Não informado','nan':'Não informado','None':'Não informado','<NA>':'Não informado','NULL':'Não informado'})

def aplicar_revisao(pasta, pacote):
    pasta=Path(pasta)
    obrigatorios=['dicionario_colunas.csv','definicoes_pendentes.csv','status.json','inventario_origens.json','validacao.csv','LEIA_ME.md']
    for nome in obrigatorios:
        if not (pasta/nome).is_file():raise FileNotFoundError(f'Arquivo necessário ausente: {pasta/nome}')
    if not (pasta/'dados').is_dir():raise FileNotFoundError('Pasta dados ausente.')
    status=json.loads((pasta/'status.json').read_text(encoding='utf-8'))
    validacao=ler_csv(pasta/'validacao.csv')
    if status.get('divergencias')!=0 or status.get('falhas') or status.get('conjuntos_exportados')!=9:
        raise ValueError('Esta revisão se destina ao produto de nove conjuntos sem falhas informado na conversa.')
    if not validacao.confere.str.lower().eq('true').all():raise ValueError('A validação de origem possui divergências.')
    atual=ler_csv(pasta/'dicionario_colunas.csv')
    pendentes=ler_csv(pasta/'definicoes_pendentes.csv')
    original=pd.DataFrame(pacote['registros_originais'])
    chaves=['conjunto','coluna']
    if atual.duplicated(chaves).any() or pendentes.duplicated(chaves).any():raise ValueError('Dicionário com chaves duplicadas.')
    cols=list(original.columns)
    def ordenar(df):return df[cols].sort_values(chaves).reset_index(drop=True).astype(str)
    if not ordenar(pendentes).equals(ordenar(original)):
        raise ValueError('As pendências desta pasta diferem do arquivo de 303 entradas revisado. A revisão não foi aplicada.')
    if not set(map(tuple,original[chaves].values)).issubset(set(map(tuple,atual[chaves].values))):
        raise ValueError('Há entradas revisadas que não constam no dicionário completo.')
    inventario=json.loads((pasta/'inventario_origens.json').read_text(encoding='utf-8'))
    if len(inventario)!=9:raise ValueError('Inventário não contém nove conjuntos.')
    dados_mapeados={}
    for item in inventario:
        nome=item['arquivo_saida']
        if Path(nome).name!=nome:raise ValueError('Nome de arquivo inesperado no inventário.')
        p=pasta/'dados'/nome
        if hash_arquivo(p)!=item['sha256_saida']:raise ValueError(f'Conteúdo de {nome} diverge do inventário. Nada foi alterado.')
        dados_mapeados[item['conjunto']]=p
    regras=pd.DataFrame(pacote['revisoes'])
    testes=[]
    for k,grupo in regras[regras.teste_derivacao.ne('')].groupby('conjunto'):
        df=ler_csv(dados_mapeados[k])
        for i,r in grupo.iterrows():
            campo,fonte,modo=r.coluna,r.coluna_derivadora,r.teste_derivacao
            resultado=True;obs=''
            if campo not in df or (modo!='constante' and fonte not in df):
                resultado=False;obs='Campo auxiliar ou coluna derivadora ausente.';n=-1
            elif modo=='constante':
                n=int((~df[campo].eq('Não informado')).sum());resultado=n==0
            elif modo=='numero':
                a=pd.to_numeric(df[campo].str.replace(',','.',regex=False),errors='coerce')
                b=pd.to_numeric(df[fonte].str.replace(',','.',regex=False),errors='coerce')
                if campo=='_ANO':b=b.where(b.between(2000,2100))
                ok=(a.eq(b) | (a.isna() & b.isna()))
                # Não aceitar erros de conversão como concordância de dois valores ausentes.
                inval=(df[campo].str.strip().ne('') & a.isna()) | (df[fonte].str.strip().ne('') & pd.to_numeric(df[fonte].str.replace(',','.',regex=False),errors='coerce').isna())
                n=int((~ok|inval).sum());resultado=n==0
            else:
                esperado=texto_limpo(df[fonte])
                if campo=='_TI_NOME':
                    mascara=esperado.eq('Não informado')
                    esperado.loc[mascara]='Terra Indígena sem nome - ID '+texto_limpo(df.loc[mascara,'_TI_ID'])
                n=int(df[campo].ne(esperado).sum());resultado=n==0
            if resultado:
                regras.at[i,'situacao_definicao']='definição documentada e derivação conferida no CSV'
                regras.at[i,'criterio_revisao']+='; igualdade conferida nesta execução'
            else:
                regras.at[i,'situacao_definicao']='definição pendente'
                regras.at[i,'observacao_revisao']+=' A derivação indicada não foi confirmada no CSV desta execução. '+obs
            testes.append({'conjunto':k,'coluna':campo,'coluna_derivadora':fonte or 'constante Não informado','linhas_divergentes':n,'confere':resultado,'observacao':obs})
    # Atualiza somente os campos documentais das 303 entradas, sem mexer em contagens ou dados.
    editaveis=['descricao','tipo_logico','unidade','situacao_definicao','fonte_definicao','criterio_revisao','observacao_revisao','coluna_derivadora','teste_derivacao']
    for c in editaveis:
        if c not in atual:atual[c]=''
    indice={tuple(r):i for i,r in enumerate(atual[chaves].values)}
    for r in regras.to_dict('records'):
        ix=indice[(r['conjunto'],r['coluna'])]
        for c in editaveis:atual.at[ix,c]=r[c]
    remanescentes=atual[atual.situacao_definicao.eq('definição pendente')].copy()
    alertas=regras[regras.observacao_revisao.ne('') & regras.teste_derivacao.ne('')].copy()
    stamp=datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%S_%fZ')
    nova=pasta.with_name(pasta.name+'_documentacao_'+stamp)
    shutil.copytree(pasta,nova)
    # Documentos substituídos ficam também em subpasta de histórico.
    historico=nova/'documentacao_anterior';historico.mkdir(exist_ok=False)
    for nome in ['dicionario_colunas.csv','definicoes_pendentes.csv','status.json','LEIA_ME.md']:
        shutil.copy2(nova/nome,historico/nome)
    for nome,df in [('dicionario_colunas.csv',atual),('definicoes_pendentes.csv',remanescentes),('revisao_303_definicoes.csv',regras),('conferencia_derivacoes.csv',pd.DataFrame(testes)),('alertas_campos_auxiliares.csv',alertas)]:
        df.to_csv(nova/nome,sep=';',index=False,encoding='utf-8-sig')
    diverg=sum(not r['confere'] for r in testes)
    revisadas=int((regras.situacao_definicao!='definição pendente').sum())
    status.update({'definicoes_pendentes':len(remanescentes),'revisao_documental_utc':stamp,
                   'entradas_documentadas_nesta_revisao':revisadas,'derivacoes_conferidas':len(testes),
                   'divergencias_derivacao':diverg,'alertas_campos_auxiliares':len(alertas),
                   'status':'DOCUMENTACAO_REVISADA_COM_PENDENCIAS_E_ALERTAS',
                   'observacao_status':'As 89 conferências de origem são preservadas. Os alertas de interpretação são distintos das contagens. Produto ainda requer revisão documental.',
                   'dados_modificados':False,'intersecoes_recalculadas':False})
    (nova/'status.json').write_text(json.dumps(status,ensure_ascii=False,indent=2),encoding='utf-8')
    nota='''\n\n## Revisão documental posterior\n
As definições foram confrontadas com Base04 e os notebooks de interseção e relatórios da versão 2.
As fontes e os critérios estão registrados em cada linha revisada do dicionário. Permanecem pendentes
os atributos cuja semântica, domínio ou unidade não foi confirmada. Proveniência conhecida não equivale
a definição semântica confirmada. Consulte definicoes_pendentes.csv.

Os campos auxiliares _TIPOLOGIA_* e _UF_UNIDADE foram documentados pela coluna efetivamente
selecionada nos códigos, com confronto adicional no CSV leve. A seleção automática por nome
parcial pode produzir dimensões com rótulo inadequado. Consulte alertas_campos_auxiliares.csv
antes de interpretar resultados agrupados. A revisão não corrige retrospectivamente esses valores,
nem altera resultados, gráficos ou textos da dissertação. Revisar os trechos que utilizem essas dimensões.
_ANO pode representar ANO_BASE em uma camada e ANO_OPERACAO em outra; o dicionário explicita isso.

Os dados foram copiados byte a byte e confrontados com os hashes do inventário. Nenhuma interseção
foi refeita. Os arquivos anteriores de documentação estão em documentacao_anterior. As 89 verificações
de origem continuam registradas em validacao.csv; os testes adicionais estão em conferencia_derivacoes.csv.
Nenhum teste de contagem confirma significado dos campos, incidência normativa ou irregularidade.
'''
    with (nova/'LEIA_ME.md').open('a',encoding='utf-8') as f:f.write(nota)
    for item in inventario:
        if hash_arquivo(nova/'dados'/item['arquivo_saida'])!=item['sha256_saida']:
            raise RuntimeError('Cópia dos dados divergiu. ZIP não produzido; pasta original intacta.')
    log={'origem':str(pasta),'destino':str(nova),'pendencias_antes':len(original),'pendencias_depois':len(remanescentes),
         'revisadas':revisadas,'fontes_notebooks':pacote['fontes_notebooks'],'fontes_intersecoes':pacote['fontes_intersecoes'],
         'dados_preservados':True,'sha256_anexo_recebido':pacote['origem_sha256']}
    (nova/'registro_revisao.json').write_text(json.dumps(log,ensure_ascii=False,indent=2),encoding='utf-8')
    zpath=None
    if diverg==0:
        zpath=Path(str(nova)+'.zip')
        with zipfile.ZipFile(zpath,'w',zipfile.ZIP_DEFLATED,compresslevel=6) as z:
            for p in sorted(nova.rglob('*')):
                if p.is_file() and not p.name.endswith('.part'):z.write(p,p.relative_to(nova))
    print(json.dumps(status,ensure_ascii=False,indent=2))
    print('Nova pasta:',nova)
    print('ZIP:',zpath or 'Não gerado: há divergências nas derivações, que precisam ser verificadas.')
    return nova,zpath
```

## Aplicar a revisão em uma cópia do produto
O ZIP novo só é produzido se os hashes dos dados e as derivações conferirem. Uma divergência de derivação mantém a definição pendente e impede o ZIP; a pasta original permanece intacta. O novo status preserva as 89 conferências anteriores separadamente dos testes documentais.


```python
PASTA_REVISADA, ZIP_REVISADO = aplicar_revisao(PASTA_ORIGINAL, PACOTE_REVISAO)
```

As entradas ainda pendentes continuam no dicionário e em uma lista própria. As 149 pendências previstas podem aumentar se uma derivação não conferir com os dados efetivos desta execução. Uma nova pasta/ZIP não significa produto documentalmente concluído.
